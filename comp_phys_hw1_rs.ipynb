```python
# Comp. Physics Integral Code
# Ryan Saelens
# 1/31/24

# First Run: N = 100 Slices

import time
import math

N = 100

def Integral(N):
    I = 0
    h = 2/N
    for k in range(1, N+1): # This is the Riemann Def. 
        xk = -1 + h * k
        yk = 1 - xk ** 2
        I = I + h * ((yk) ** .5)
    return I
    
I = Integral(100)

print("When N = 100, the value of the integral is:", I)
print("The Absolute Difference:", abs(I - .5 * math.pi))
    
```

```python
# Continued Code


# To time the execution of the code
import timeit

# loop to increase the size of N --> 10^4 , 10^5 , 10^6
for i in [4, 5, 6]:
    N = 10 ** i
    tic = timeit.default_timer()
    I = Integral(N)
    toc = timeit.default_timer()
    print("N =", N)
    print("I =", I)
    print("The Absolute Difference:", abs(I - .5 * math.pi))
    print("Time In Seconds: {:g} sec".format(toc - tic))
```
