```python
# comp-phys examples from class


# Example 6.1: Gaussian Elim. W/ Backsubstitution


from numpy import array, empty
A = array ([[2,1,4,1],
          [3,4,-1,-1],
          [1,-4,1,5],
          [2,-2,1,3]],float)
v = array([-4,3,9,7],float)
N = len(v)

#Solve Ax = v --> find the unknown column vector x

for m in range(N):
    # divide by diagonal element first to set equal to 1
    div_elem = A[m,m] # top-left corner of matrix :: if equal to zero, things will break!
    A[m,:] /= div_elem # divides row by div_elem
    v[m] /= div_elem # divide v array to keep eqn the same
    
    # first row done --> should have 1 in top-left corner
    # now we want zeros underneath the diagonal element, which will grow with each new row being computed
    # therefore we should end up with a diagonal of 1's and zeros beneath them
    
    for i in range(m+1, N):
        mult_elem = A[i,m] # for each row i below above one, column stays the same
        A[i, :] -= mult_elem * A[m,:] #subtract from new row: mult-factor*the first row
        v[i] -= mult_elem * v[m] # same is done for v to make sure things stay the same
        
# perform backsubstitution 
x = empty(N,float)
for m in range(N-1, -1, -1): # go in reverse order
    x[m] = v[m] # great --> x = v
    for i in range(m+1, N): # must update remaining pieces
        x[m] -= A[m,i] * x[i]
        
print(x)
```
